```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("student_performance.csv")

df.head()
```

```python
df.isnull().sum()
```

```python
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())

df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].mean())

df["Quiz_Score"] = df["Quiz_Score"].fillna(df["Quiz_Score"].mean())
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()

df["Study_Hours_Normalized"] = scaler.fit_transform(df[["Study_Hours"]])

df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Gender_Encoded"] = encoder.fit_transform(df["Gender"])

df.head()
```

```python
scaler = MinMaxScaler()

df["Attendance_Normalized"] = scaler.fit_transform(df[["Attendance"]])

df.head()

```

```python
df.head()
```

```python
df.to_csv("student_performance_cleanedd.csv", index=False)
```
