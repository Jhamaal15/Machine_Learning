```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_performance.csv")
df.head()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df["Final_Score"][:15], marker="o")

plt.title("Final Score Of First 15 Students")
plt.xlabel("Student Index")
plt.ylabel("Final Score")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

avg_scores = df.groupby("Department")["Final_Score"].mean()

plt.bar(avg_scores.index, avg_scores.values, color="skyblue")

plt.title("Average Final Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Final Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Score"
)

plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Study_Hours"], bins=10)
plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")
plt.show()
```

```python
correlation = df.select_dtypes(include="number").corr()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")
plt.show()
```

#Insights

Looking at the line chart, final scores go up and down a lot between students, so there's no clear pattern based on order. The bar chart shows most departments have similar average scores, but Software Engineering students score a little higher than the rest. In the scatter plot, students who study more hours tend to get higher final scores, showing study hours really do matter. The heatmap backs this up, showing a strong link between study hours and final score, though exam, quiz, and assignment scores matter even more. Overall, how well a student does on their tests and how much they study seem to matter a lot more than their department or age.
